```python
orders = [{"id": 2001,"customer": "Anan", "city": "Korat","product":"Keyboard", "qty": 2, "price": 450},
          {"id": 2002,"customer": "Benja", "city": "Bangkok","product":"Mouse",    "qty": 5,"price": 180},
          {"id": 2003,"customer": "Chai", "city": "Bangkok","product":"Monitor",  "qty": 1,"price": 3200},
          {"id": 2004,"customer": "Duang", "city": "Phuket","product":"Mouse",    "qty": 3,"price": 180},
          {"id": 2005,"customer": "Anan", "city": "Korat","product":"Monitor",  "qty": 2,"price": 3200},
          {"id": 2006,"customer": "Ekk",  "city": "Bangkok","product":"Keyboard", "qty": 4, "price": 450},
          ]
sales = []
total_sales = 0
for order in orders:
    order_id = order["id"]
    subtotal = order["qty"] * order["price"]
    vat = subtotal * 0.07
    net_total = subtotal + vat
    sales.append((order_id, subtotal, net_total))
    total_sales += net_total

print("Order ID  Subtotal  Total VAT")
for sale in sales:
    print(f"{sale[0]}  {sale[1]:,.2f}  {sale[2]:,.2f}")
print(f"Total Sales = {total_sales:,.2f} Baht")
```
